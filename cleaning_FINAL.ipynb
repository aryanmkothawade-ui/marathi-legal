```python
# Ready-to-Use OCR Notebook for Marathi PDFs (with Tesseract LSTM Best Model)
# --------------------------------------------------------------
# This notebook shows:
# 1. How to install dependencies (Colab + Local instructions)
# 2. How to render PDF pages as images using PyMuPDF
# 3. How to run OCR with Tesseract using LSTM "best" model
# 4. How to save extracted text
# --------------------------------------------------------------

# ============================
# SECTION 1 — INSTALLATIONS
# ============================
# If running in Google Colab uncomment:
!apt-get install -y tesseract-ocr
!pip install pytesseract
!pip install pymupdf
!pip install pillow

# For local Jupyter, install manually:
# pip install pytesseract pymupdf pillow

import pytesseract
import fitz   # PyMuPDF
from PIL import Image, ImageOps, ImageFilter
import os

# OPTIONAL: If using Windows, set Tesseract path
# pytesseract.pytesseract.tesseract_cmd = r"C:/Program Files/Tesseract-OCR/tesseract.exe"

# OPTIONAL: If using custom LSTM best model, set directory
os.environ['TESSDATA_PREFIX'] = r"/content"  # folder containing mar.traineddata

# ============================
# SECTION 2 — PDF → IMAGE
# ============================

def pdf_to_images(pdf_path, dpi=300):
    """
    Converts each PDF page to a PIL image at high DPI (recommended for OCR).
    """
    doc = fitz.open(pdf_path)
    images = []
    for i, page in enumerate(doc):
        mat = fitz.Matrix(dpi/72, dpi/72)
        pix = page.get_pixmap(matrix=mat, alpha=False)

        img = Image.frombytes("RGB", [pix.width, pix.height], pix.samples)
        images.append(img)
    doc.close()
    return images

# ============================
# SECTION 3 — IMAGE PREPROCESSING FOR OCR
# ============================

def preprocess_image(img):
    """
    Basic enhancements: grayscale, autocontrast, sharpen, threshold.
    """
    gray = ImageOps.grayscale(img)
    gray = ImageOps.autocontrast(gray)
    gray = gray.filter(ImageFilter.SHARPEN)

    # Simple threshold
    gray = gray.point(lambda x: 0 if x < 128 else 255, '1')
    return gray

# ============================
# SECTION 4 — OCR FUNCTION (using LSTM best model)
# ============================

def ocr_image(img, tessdata_dir=None):
    """
    OCR using Tesseract LSTM engine.
    oem=1 → LSTM ONLY
    psm=6 → Assume uniform block of text
    """
    config = "--oem 1 --psm 6"
    # Removed the tessdata_dir argument from config as TESSDATA_PREFIX is already set.
    # if tessdata_dir:
    #     config += f" --tessdata-dir '{tessdata_dir}'"

    return pytesseract.image_to_string(img, lang='mar', config=config)

# ============================
# SECTION 5 — MAIN PIPELINE
# ============================

def extract_text_from_pdf(pdf_path, tessdata_dir=None):
    images = pdf_to_images(pdf_path)
    final_text = []

    for i, img in enumerate(images):
        print(f"Processing page {i+1}/{len(images)}...")
        proc = preprocess_image(img)
        # tessdata_dir is passed to ocr_image, but ocr_image will now rely on os.environ['TESSDATA_PREFIX']
        text = ocr_image(proc, tessdata_dir=tessdata_dir)
        final_text.append(text)

    return "\n".join(final_text)

# ============================
# SECTION 6 — USAGE EXAMPLE
# ============================
pdf_path = "/content/3.pdf"
result = extract_text_from_pdf(pdf_path, tessdata_dir="/content")

# Save output
with open("output_marathi_ocr2.txt", "w", encoding="utf-8") as f:
     f.write(result)

print("OCR Complete. Output saved to output_marathi_ocr2.txt")
```

```python
import mwxml
import bz2
import re
from collections import Counter

counter = Counter()
dev_re = re.compile(r'[\u0900-\u097F]+')

dump_path = "mrwiki-20251120-pages-articles.xml.bz2"

with bz2.open(dump_path) as f:
    dump = mwxml.Dump.from_file(f)

    for page in dump:
        for revision in page:
            text = revision.text or ""
            words = dev_re.findall(text)
            counter.update(words)

# --- Write output file ---
with open("marwiki_word_freq.txt", "w", encoding="utf-8") as out:
    for word, freq in counter.most_common():
        out.write(f"{word}\t{freq}\n")

print("Done! Total unique words:", len(counter))

# --- Print first few lines ---
print("\nSample from marwiki_word_freq.txt:")
with open("marwiki_word_freq.txt", "r", encoding="utf-8") as f:
    for i, line in enumerate(f):
        if i == 30:     # print first 10 lines
            break
        print(line.strip())
```

```python
from indicnlp.normalize.indic_normalize import IndicNormalizerFactory

normalizer = IndicNormalizerFactory().get_normalizer("mr")
print("Normalizer loaded!")
```

```python
text = open("output_marathi_ocr2.txt", "r", encoding="utf-8").read()
normalized_text = normalizer.normalize(text)

print(normalized_text[:500])
```

```python
from symspellpy import SymSpell, Verbosity

sym = SymSpell(max_dictionary_edit_distance=2, prefix_length=7)
sym.load_dictionary("/content/marwiki_word_freq.txt", term_index=0, count_index=1)
```

```python
def symspell_correct(text):
    result = []
    for word in text.split():
        suggestion = sym.lookup(word, Verbosity.CLOSEST, max_edit_distance=2)
        if suggestion:
            result.append(suggestion[0].term)
        else:
            result.append(word)
    return " ".join(result)

corrected_text = symspell_correct(normalized_text)
```

```python
with open("marathi_corrected.txt", "w", encoding="utf-8") as f:
    f.write(corrected_text)
```

```python
import re

# Step 4: Fix number segmentation
fixed_numbers_text = re.sub(r'(?<=\d)\s+(?=\d)', '', corrected_text)

# Optional: fix slash-separated numbers
fixed_numbers_text = re.sub(r'(?<=\d)\s*/\s*(?=\d)', '/', fixed_numbers_text)

# Save final output
with open("marathi_corrected_final.txt", "w", encoding="utf-8") as f:
    f.write(fixed_numbers_text)

print("Saved cleaned file as marathi_corrected_final.txt")
```

```python
def merge_paragraphs(text):
    lines = text.split("\n")
    merged = []
    buffer = ""

    # These characters indicate a line *can end* here
    end_marks = ("।", ".", "?", "!", ")", "”", "’")

    for line in lines:
        line = line.strip()
        if not line:
            continue

        if buffer == "":
            buffer = line
        else:
            # if previous line does not end with punctuation → merge
            if not buffer.endswith(end_marks):
                buffer += " " + line
            else:
                merged.append(buffer)
                buffer = line

    # add final leftover
    if buffer:
        merged.append(buffer)

    return "\n\n".join(merged)
```

```python
with open("marathi_corrected_final.txt", "r", encoding="utf-8") as f:
    text_after_step4 = f.read()

paragraph_restored = merge_paragraphs(text_after_step4)
```

```python
def clean_text(text):
    # remove extra spaces
    text = re.sub(r"\s+", " ", text)

    # remove space before punctuation
    text = re.sub(r"\s+([।.,?!])", r"\1", text)

    # restore paragraph line breaks properly
    text = text.replace(" ।", "।")

    return text.strip()
```

```python
final_clean_text = clean_text(paragraph_restored)
```
